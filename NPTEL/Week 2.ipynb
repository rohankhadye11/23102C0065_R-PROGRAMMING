## Knowing numbers and characters

```python
%%R
x=20

print(is.numeric(x))

print(is.character(x))
```

```python
%%R
m="apple"

print(is.numeric(m))

print(is.character(m))
```

## Converting numbers and characters

```python
%%R
x = 20
print(is.numeric(x))

y = as.character(x)
print(is.numeric(y))

print(is.character(y))

print(y)
```

```python
%%R
y = "apple"
print(is.numeric(y))
print(is.character(y))

z = suppressWarnings(as.numeric(y))
print(is.numeric(z))

print(is.character(z))

print(z)
```
