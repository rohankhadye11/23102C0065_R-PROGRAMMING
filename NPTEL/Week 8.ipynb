## Lists – Creating and Merging Lists

```r
list1 = list(1, 2, 3)
list2 = list("water", "juice", "lemonade")

list1
list2

list12 = c(list1, list2)
list12
```

## Lists – Converting Lists to Vectors

```r
list1 = list(1, 2, 3)
list2 = list("water", "juice", "lemonade")

unlist(list1)
unlist(list2)

mode(list1)
mode(unlist(list1))
```

## Lists – Appending Elements

```r
list1 = list(1, 2, 3)
list2 = list("water", "juice", "lemonade")

append(list1, 100)
append(list2, "coffee")
```

## Lists – Appending After a Position

```r
list1 = list(1, 2, 3)
list2 = list("water", "juice", "lemonade")

append(list1, 100, after = 2)
append(list2, "coffee", after = 2)
```

## Lists – Removing Elements by Position

## Lists – Extracting Elements by Index

```r
list1 = list(1, 2, 3, 4, 5, 6)
list2 = list("water", "juice", "lemonade", "tea", "coffee", "milk")

list1[2:4]
list1[c(1, 3, 5)]

list2[2:4]
list2[c(1, 3, 5)]
```

## Vector Indexing – Conditional Indexing

```r
x = 1:10
x

x[(x > 5)]

x[(x %% 2 == 0)]  # values for which x mod 2 is 0
x[(x %% 2 == 1)]  # values for which x mod 2 is 1
```

## Vector Indexing – Missing Observations

```r
x = 1:10
x[5] = NA
x

y = x[!is.na(x)]  # ! means negation
y

mean(x)
mean(y)
```

## Vector Indexing – Negative Integer Indexing

```r
x = 1:10
x

x[-(1:5)]

x[6:10]
```

## Factors – Creating a Factor

```r
y = c(1, 4, 3, 5, 4, 2, 4)
possible.dieface = c(1, 2, 3, 4, 5, 6)
labels.dieface = c("one", "two", "three", "four", "five", "six")

facy = factor(
  y,
  levels = possible.dieface,
  labels = labels.dieface
)

facy
```

## Factors – as.factor

```r
x = c(3, 4, 5, 6, 1, 2, 3, 3, 4, 4, 5, 6)
x

y = as.factor(x)
y

x = factor(c("lemonade", "lemonade", "juice", "lemonade", "water"))
x
```

## Factors – Class Function

```r
class(9)
class("9")
class(print)

x = matrix(nrow = 2, ncol = 2, data = 1:4)
class(x)
```

## Factors – Unclass Function

```r
brands = c("A", "A", "B", "B", "B", "B", "C")
brands

brands_fac = factor(brands)
brands_fac

unclass(brands_fac)

colours = c("blue", "green", "red")
colours

colours[unclass(brands_fac)]
```

## Factors – Custom Factor Levels

```r
x = factor(
  c("lemonade", "lemonade", "juice", "lemonade", "water"),
  levels = c("water", "juice", "lemonade")
)

x
unclass(x)
levels(x)
```

## Factors – Ordered Factors

```r
income = ordered(
  c("high", "high", "low", "medium", "medium"),
  levels = c("low", "medium", "high")
)

income
unclass(income)
```

## Strings – Format Function

```r
format(0.5, digits = 10, nsmall = 15)

format(
  c("A", "BB", "CCC", "DDDD"),
  width = 7,
  justify = "centre"
)

format(
  c("A", "BB", "CCC", "DDDD"),
  width = 14,
  justify = "centre"
)
```

## Strings – Format with Justification

```r
format(c("A", "BB", "CCC", "DDDD"), width = 7, justify = "centre")
format(c("A", "BB", "CCC", "DDDD"), width = 7, justify = "left")
format(c("A", "BB", "CCC", "DDDD"), width = 7, justify = "right")
format(c("A", "BB", "CCC", "DDDD"), width = 7, justify = "none")
```
